Punto 1

```python
# ==============================
# MEZCLA DE FERTILIZANTES
# ==============================

import numpy as np
from scipy.optimize import linprog

# Funcion objetivo (negativa porque linprog minimiza)
c = [-180, -220]

# Restricciones
A = [
    [3, 4],
    [2, 1],
    [5, 3]
]

b = [120, 80, 150]

# Limites
bounds = [(0, None), (0, None)]

# ==============================
# RESOLVER
# ==============================

resultado = linprog(c, A_ub=A, b_ub=b, bounds=bounds, method='highs')

x1, x2 = resultado.x

# ==============================
# RESULTADOS
# ==============================

print("=" * 50)
print("MEZCLA DE FERTILIZANTES")
print("=" * 50)

print(f"Estado: {'Optimo encontrado' if resultado.success else 'Error'}")
print()

print(f"Fertilizante F1 (x1): {x1:.2f}")
print(f"Fertilizante F2 (x2): {x2:.2f}")

ganancia = -resultado.fun
print(f"\nGanancia maxima: ${ganancia:.2f}")

print("=" * 50)

# ==============================
# VERIFICACION
# ==============================

print("\nVERIFICACION:")

print(f"A: {3*x1 + 4*x2:.2f} <= 120")
print(f"B: {2*x1 + x2:.2f} <= 80")
print(f"C: {5*x1 + 3*x2:.2f} <= 150")
```

```python
# ==============================
# PRODUCCION SEMANAL DE UNA FABRICA
# ==============================

import numpy as np

# Datos iniciales
S0 = 200
M0 = 80

# Matriz del sistema
A = np.array([
    [0.6, 0.2],
    [0.1, 0.5]
])

b = np.array([40, 20])

x0 = np.array([S0, M0])

# ==============================
# CALCULO DE SEMANAS
# ==============================

x1 = A @ x0 + b
x2 = A @ x1 + b

# ==============================
# EQUILIBRIO
# ==============================

equilibrio = np.linalg.solve(np.eye(2) - A, b)

# ==============================
# RESULTADOS
# ==============================

print("=" * 50)
print("PRODUCCION SEMANAL")
print("=" * 50)

print(f"Semana 0 -> Sillas = {x0[0]:.2f}, Mesas = {x0[1]:.2f}")
print(f"Semana 1 -> Sillas = {x1[0]:.2f}, Mesas = {x1[1]:.2f}")
print(f"Semana 2 -> Sillas = {x2[0]:.2f}, Mesas = {x2[1]:.2f}")

print("\nEQUILIBRIO:")
print(f"Sillas = {equilibrio[0]:.2f}")
print(f"Mesas = {equilibrio[1]:.2f}")

print("=" * 50)

# ==============================
# VERIFICACION
# ==============================

verif = A @ equilibrio + b

print("\nVERIFICACION:")
print(f"S* = {equilibrio[0]:.2f} | RHS = {verif[0]:.2f}")
print(f"M* = {equilibrio[1]:.2f} | RHS = {verif[1]:.2f}")
```

```python
# ==============================
# EPIDEMIA EN UNA CIUDAD - SIS
# ==============================

import numpy as np

# Datos
N = 10000
beta = 0.00003
gamma = 0.1

I0 = 50
S0 = N - I0

# ==============================
# FUNCION DEL MODELO
# ==============================

def dI_dt(I):
    return beta * (N - I) * I - gamma * I

# ==============================
# EQUILIBRIOS
# ==============================

I_eq_1 = 0.0
I_eq_2 = N - (gamma / beta)

S_eq_1 = N - I_eq_1
S_eq_2 = N - I_eq_2

# ==============================
# RESULTADOS
# ==============================

print("=" * 50)
print("EPIDEMIA EN UNA CIUDAD - MODELO SIS")
print("=" * 50)

print("VARIABLES:")
print("I(t): infectados")
print("S(t): susceptibles")
print()

print("CLASIFICACION:")
print("No lineal - continuo")
print()

print(f"Condicion inicial: I(0) = {I0}")
print(f"S(0) = {S0}")
print()

print("EQUILIBRIOS:")
print(f"I* = {I_eq_1:.2f}, S* = {S_eq_1:.2f}")
print(f"I* = {I_eq_2:.2f}, S* = {S_eq_2:.2f}")

print("=" * 50)

# ==============================
# VERIFICACION
# ==============================

print("\nVERIFICACION:")

print(f"dI/dt en I* = {I_eq_1:.2f}: {dI_dt(I_eq_1):.6f}")
print(f"dI/dt en I* = {I_eq_2:.2f}: {dI_dt(I_eq_2):.6f}")

print("\nINTERPRETACION:")
print("I* = 0 -> la enfermedad desaparece")
print("I* alto -> estado endemico")
```

```python
# ==============================
# MODELO LOGISTICO DE PESCA
# ==============================

import numpy as np

# --- DATOS DEL PROBLEMA ---
r = 0.4      # tasa de crecimiento
K = 1000     # capacidad maxima
H = 80       # pesca anual
P0 = 400     # poblacion inicial

# ==============================
# MODELO DISCRETO
# ==============================

def siguiente_poblacion(P):
    return P + r * P * (1 - P / K) - H

# ==============================
# CALCULO DE P1 Y P2
# ==============================

P1 = siguiente_poblacion(P0)
P2 = siguiente_poblacion(P1)

# ==============================
# PUNTOS DE EQUILIBRIO
# ==============================

# r*P*(1 - P/K) - H = 0
coeficientes = [r / K, -r, H]

equilibrios = np.roots(coeficientes)
equilibrios = np.sort(np.real_if_close(equilibrios))

P_eq_1 = float(equilibrios[0])
P_eq_2 = float(equilibrios[1])

# ==============================
# ESTABILIDAD
# ==============================

def derivada_modelo(P):
    return 1 + r - (2 * r * P / K)

der_eq_1 = derivada_modelo(P_eq_1)
der_eq_2 = derivada_modelo(P_eq_2)

estabilidad_1 = "estable" if abs(der_eq_1) < 1 else "inestable"
estabilidad_2 = "estable" if abs(der_eq_2) < 1 else "inestable"

# ==============================
# RESULTADOS
# ==============================

print("=" * 50)
print("MODELO LOGISTICO DE PESCA")
print("=" * 50)

print("VARIABLES Y PARAMETROS:")
print("Variable: P_n")
print(f"r = {r}")
print(f"K = {K}")
print(f"H = {H}")

print("\nCLASIFICACION:")
print("No lineal - discreto")

print(f"\nPoblacion inicial: P0 = {P0:.2f}")
print(f"P1 = {P1:.2f}")
print(f"P2 = {P2:.2f}")

print("\nEQUILIBRIOS:")
print(f"P*1 = {P_eq_1:.2f} ({estabilidad_1})")
print(f"P*2 = {P_eq_2:.2f} ({estabilidad_2})")

print("=" * 50)

# ==============================
# VERIFICACION
# ==============================

print("\nVERIFICACION:")
print(f"F(P*) en {P_eq_1:.2f} = {siguiente_poblacion(P_eq_1):.2f}")
print(f"F(P*) en {P_eq_2:.2f} = {siguiente_poblacion(P_eq_2):.2f}")

print("\nINTERPRETACION:")
print("Si la poblacion inicia por debajo del equilibrio inestable, colapsa.")
print("Si inicia por encima, tiende al equilibrio estable.")
```
